Compute Mandelbrot escape counts in pure Python, split the image into independent row tiles, and compare identical results across AGILAB execution modes. This is a CPU benchmark, not a claim that Python loops are the fastest way to render fractals. Reference: https://docs.python.org/3.14/howto/free-threading-python.html


```python
def reference_image(width=96, height=64, iterations=100):
    pixels = []
    for y in range(height):
        cy = -1.2 + 2.4 * y / (height - 1)
        for x in range(width):
            cx = -2.0 + 3.0 * x / (width - 1)
            z = 0j
            c = complex(cx, cy)
            count = 0
            while count < iterations and z.real*z.real + z.imag*z.imag <= 4.0:
                z = z*z + c
                count += 1
            pixels.append(count)
    return pixels
```

```python
pixels = reference_image()
assert len(pixels) == 96 * 64
assert min(pixels) >= 0 and max(pixels) <= 100
```
