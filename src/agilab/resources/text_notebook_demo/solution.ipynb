# Text Atlas – Dimensionality Reduction of Text Data

Adapted from the INRIA scikit-learn MOOC notebook *Dimensionality reduction of
text data* (CC BY 4.0, INRIA scikit-learn MOOC contributors).

Corpus: 1,250 Wikinews articles (CC BY 2.5, Wikinews contributors, curated by
Mega Rhyme).

This notebook loads the corpus, runs the TF-IDF → PCA(50) → KMeans pipeline,
and writes `results.json` with all computed fields.

```python
import json
from pathlib import Path
import numpy as np
import pandas as pd

from text_core import load_corpus, analyze
```

```python
# Stage 1: Load corpus
df = load_corpus()
print(f"Loaded {len(df)} documents, {df['category'].nunique()} categories")
print(df.head())

# Stage 2: Run analysis
result = analyze()
print(f"Vocabulary: {len(result['vocabulary'])} terms")
print(f"Silhouette (50-D): {result['silhouette']:.4f}")
print(f"Displayed variance (2-D): {result['displayed_variance']:.4f}")
print(f"Retained variance (50-D): {result['retained_variance']:.4f}")

# Stage 3: Write results.json to current execution directory
results = {
    "results": {
        "coordinates": result["coordinates"].tolist(),
        "labels": result["labels"].tolist(),
        "vocabulary": result["vocabulary"],
        "top_terms": result["top_terms"],
        "silhouette": result["silhouette"],
        "displayed_variance": result["displayed_variance"],
        "retained_variance": result["retained_variance"],
        "parameters": result["parameters"],
    }
}

out_path = Path("results.json")
out_path.write_text(json.dumps(results, indent=2))
print(f"Wrote {out_path}")
```
