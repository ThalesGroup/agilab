This notebook trains a Decision Tree, Random Forest, and Logistic Regression on the Iris dataset using a stratified 70/30 split, evaluates them on the held-out test set, and writes `metrics.json` with per-model accuracy.

> **Caveat:** Tuning hyper-parameters (e.g. tree depth) on the displayed holdout makes results *exploratory*. Final performance claims require a separate, untouched test set.

```python
import json
import numpy as np
import pandas as pd
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score
import models

# --- Import & Prepare ---
iris = load_iris(as_frame=True)
X = iris.data  # all four features
y = iris.target

# --- Train ---
seed = 42
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.3, stratify=y, random_state=seed
)

model_dict = models.build_models(max_depth=3, seed=seed)
for name, clf in model_dict.items():
    clf.fit(X_train, y_train)

print(f"Train: {len(X_train)}  Test: {len(X_test)}")
print("Fitted:", list(model_dict.keys()))
```

```python
# --- Evaluate ---
records = []
for name, clf in model_dict.items():
    acc = float(accuracy_score(y_test, clf.predict(X_test)))
    records.append({"model": name, "accuracy": acc})
    print(f"{name}: {acc:.4f}")

with open("metrics.json", "w") as f:
    json.dump(records, f, indent=2)

print("Wrote metrics.json")
```

## Interpretation

On this 150-sample dataset all three models typically exceed 0.90 accuracy. The Decision Tree and Random Forest exploit non-linear boundaries in petal measurements; Logistic Regression captures the roughly linear separability. Because the holdout is small and hyper-parameters were chosen by inspection, treat these numbers as *exploratory*—a proper evaluation would use a separate untouched test set or cross-validation.
